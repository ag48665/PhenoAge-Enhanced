# Enhanced PhenoAge Model

## Project Aim

The aim of this project is to investigate how the original PhenoAge model could be expanded using additional clinical biomarkers, lifestyle information, and modern statistical learning methods.

The project focuses on:

- understanding the original PhenoAge model,
- identifying biologically meaningful additional biomarkers,
- designing a mathematically valid enhanced model,
- considering lifestyle-data integration,
- discussing validation, overfitting, fairness, and practical implementation.

## What is Biological Age?

Chronological age represents the number of years an individual has lived. However, people of the same chronological age often differ considerably in their health status, physiological function, and risk of developing age-related diseases.

Biological age aims to measure the functional state of the organism rather than simply counting the number of years lived. It reflects cumulative molecular and physiological changes occurring during ageing and may provide a better estimate of disease risk, frailty, and life expectancy.

Consequently, biological age has become an important concept in precision medicine and healthy ageing research.

## What is PhenoAge?

Phenotypic Age (PhenoAge) is a biological age estimator introduced by Morgan Levine and colleagues. Unlike chronological age, PhenoAge combines age with routine clinical biomarkers to estimate mortality risk and physiological ageing.

The original model was developed using data from the National Health and Nutrition Examination Survey (NHANES) and validated in independent cohorts. It predicts all-cause mortality more accurately than chronological age alone.

One of the greatest strengths of PhenoAge is that it relies on routine blood tests that are widely available in clinical practice, making it practical for population studies and personalised medicine.

## Original PhenoAge Biomarkers

| Biomarker | Biological System | Ageing Process |
|------------|-------------------|----------------|
| Albumin | Liver function | Nutrition and frailty |
| Creatinine | Kidney function | Renal ageing |
| Glucose | Metabolism | Insulin resistance |
| C-reactive protein (CRP) | Inflammation | Inflammaging |
| Lymphocyte percentage | Immune system | Immunosenescence |
| Mean Cell Volume (MCV) | Red blood cells | Haematological ageing |
| Red Cell Distribution Width (RDW) | Blood physiology | Mortality risk |
| Alkaline Phosphatase (ALP) | Liver and bone | Tissue dysfunction |
| White Blood Cell Count | Immune activation | Chronic inflammation |

# Proposed Biomarker Expansion

The original PhenoAge model is a robust predictor of biological ageing based on routine clinical biomarkers. However, advances in ageing research suggest that additional biomarkers could improve its ability to capture multiple hallmarks of ageing.

The proposed enhanced model incorporates biomarkers related to chronic inflammation, metabolic regulation, cardiovascular health, mitochondrial function, oxidative stress, and cellular senescence. The goal is to create a more comprehensive representation of biological ageing while maintaining clinical relevance.

## Proposed Additional Biomarkers

| Biomarker | Biological Process | Why Include It? |
|------------|-------------------|-----------------|
| Interleukin-6 (IL-6) | Chronic inflammation | Strong marker of inflammaging and age-related disease risk |
| Tumour Necrosis Factor-α (TNF-α) | Immune ageing | Reflects chronic immune activation |
| HbA1c | Glucose metabolism | Indicates long-term glycaemic control |
| Apolipoprotein B (ApoB) | Lipid metabolism | Better predictor of cardiovascular risk than LDL cholesterol |
| Cystatin C | Kidney function | More sensitive marker of renal ageing than creatinine |
| Growth Differentiation Factor-15 (GDF-15) | Cellular stress | Associated with mitochondrial dysfunction and frailty |
| Fibroblast Growth Factor-21 (FGF21) | Energy metabolism | Marker of metabolic adaptation and mitochondrial health |
| NT-proBNP | Cardiovascular function | Reflects cardiac ageing and heart failure risk |

## Scientific Justification

Each proposed biomarker reflects a different aspect of the ageing process.

- **IL-6** and **TNF-α** are key cytokines involved in chronic low-grade inflammation ("inflammaging"), one of the major hallmarks of ageing.

- **HbA1c** represents long-term glucose exposure and provides a more stable indicator of metabolic health than a single fasting glucose measurement.

- **ApoB** reflects the total number of atherogenic lipoprotein particles and is strongly associated with cardiovascular disease risk.

- **Cystatin C** is considered a more reliable marker of kidney function than creatinine because it is less influenced by muscle mass.

- **GDF-15** increases during cellular stress and mitochondrial dysfunction and has emerged as an important biomarker of biological ageing.

- **FGF21** regulates energy metabolism and mitochondrial adaptation, linking metabolic stress with healthy ageing.

- **NT-proBNP** reflects cardiac function and may improve prediction of cardiovascular ageing.

# Mathematical Model Design

The original PhenoAge model estimates biological age using chronological age and a set of clinical biomarkers. An enhanced model could incorporate additional biomarkers while maintaining interpretability and clinical usefulness.

The general form of the model can be written as:

\[
Biological\ Age = f(X_1, X_2, ..., X_n)
\]

where each \(X_i\) represents a biomarker associated with ageing.

## Weighted Biomarker Model

A simple linear formulation could be expressed as:

\[
BioAge =
\beta_0 +
\beta_1 Albumin +
\beta_2 Creatinine +
\beta_3 CRP +
\beta_4 HbA1c +
\beta_5 IL6 +
\beta_6 ApoB +
\beta_7 GDF15 + ...
\]

where each coefficient (\(\beta\)) represents the contribution of an individual biomarker to biological age.

## Possible Machine Learning Models

Several statistical and machine-learning approaches could be used to estimate biological age.

| Model | Advantages | Limitations |
|------|-------------|-------------|
| Linear Regression | Simple and interpretable | Cannot model complex nonlinear relationships |
| LASSO Regression | Performs automatic feature selection | May discard correlated biomarkers |
| Elastic Net | Combines LASSO and Ridge regularisation | Requires parameter tuning |
| Random Forest | Captures nonlinear interactions | Less interpretable |
| XGBoost | High predictive accuracy | More computationally intensive |
| Neural Networks | Can learn highly complex relationships | Large datasets required; limited interpretability |

## Proposed Modelling Strategy

For an enhanced PhenoAge model, Elastic Net regression would be an appropriate baseline because it can handle correlated biomarkers while reducing the risk of overfitting.

Random Forest and XGBoost could be developed as complementary models to capture nonlinear interactions among biomarkers.

The final model should be selected based on predictive performance, interpretability, and external validation rather than predictive accuracy alone.

# Lifestyle Data Integration

Lifestyle strongly influences biological ageing, but lifestyle variables are often noisy, self-reported, and sensitive to short-term behaviour. For this reason, they should be integrated carefully rather than treated as equivalent to laboratory biomarkers.

Potential lifestyle variables include:

- sleep duration,
- sleep quality,
- physical activity,
- diet quality,
- smoking,
- alcohol intake,
- perceived stress,
- body mass index,
- resting heart rate,
- heart rate variability,
- daily step count,
- wearable-derived activity measures.

## Direct Integration vs Separate Lifestyle Score

There are two possible strategies.

### Direct integration

Lifestyle variables could be added directly to the predictive model together with clinical biomarkers.

\[
Enhanced\ BioAge = f(Biomarkers,\ Lifestyle\ Variables)
\]

This approach may improve predictive accuracy, but it may also reduce interpretability and increase the risk of overfitting.

### Separate lifestyle score

A safer and more interpretable approach would be to create a separate lifestyle score and combine it with biomarker-based age.

\[
Enhanced\ BioAge =
w_1(PhenoAge) +
w_2(Lifestyle\ Score) +
w_3(Metabolic\ Score) +
w_4(Inflammation\ Score)
\]

The weights \(w_1, w_2, w_3,\) and \(w_4\) would be estimated from training data.

## Proposed Lifestyle Score

A lifestyle score could be constructed from standardised variables.

| Lifestyle factor | Health-promoting direction |
|------------------|----------------------------|
| Sleep duration | Approximately 7–9 hours |
| Sleep quality | Higher quality |
| Physical activity | Higher regular activity |
| Diet quality | Higher-quality diet |
| Smoking | No current smoking |
| Alcohol intake | Low or moderate intake |
| Stress | Lower chronic stress |
| BMI | Within a healthy range |
| Resting heart rate | Lower within the healthy range |
| Heart rate variability | Higher within age-appropriate ranges |
| Daily step count | Higher regular movement |

Each variable could be transformed into a standardised score and combined into a composite lifestyle index.

## Scientific Considerations

Lifestyle data should be averaged over time whenever possible because single measurements may not represent habitual behaviour.

Wearable-derived data may improve objectivity, but device type, adherence, measurement quality, and missing data must be considered.

Self-reported variables such as diet, sleep, alcohol intake, and stress may contain recall bias. Therefore, lifestyle information should complement rather than replace clinical biomarkers.

# Critical Evaluation

Although adding more biomarkers has the potential to improve biological age estimation, increasing model complexity also introduces several challenges. A scientifically useful model should balance predictive performance, interpretability, and clinical applicability.

## Risk of Overfitting

Including too many biomarkers may lead to overfitting, where the model learns random patterns present in the training dataset instead of general biological relationships.

Overfitted models often achieve excellent performance on training data but perform poorly when applied to new individuals.

To minimise overfitting, feature selection methods such as LASSO or Elastic Net regression can be used. Additionally, cross-validation and independent external validation datasets should be incorporated into the model development process.

## Model Validation

A proposed enhanced PhenoAge model should be evaluated using multiple validation strategies.

Recommended approaches include:

- k-fold cross-validation,
- external validation using independent cohorts,
- comparison with the original PhenoAge model,
- evaluation of prediction error (MAE, RMSE),
- survival analysis,
- calibration plots.

A model should only be considered clinically useful if it demonstrates consistent performance across different populations.

## Generalisability

Ageing is influenced by genetics, sex, ethnicity, environmental exposures, socioeconomic status, and lifestyle. Consequently, a model developed using one population may not perform equally well in another.

The enhanced model should therefore be evaluated across:

- different age groups,
- males and females,
- multiple ethnic backgrounds,
- healthy individuals,
- patients with chronic diseases.

Ensuring fairness and robustness across diverse populations is essential before clinical implementation.

## Capillary Blood Testing

Capillary blood sampling offers a convenient and minimally invasive alternative to venous blood collection. However, not all biomarkers can currently be measured with sufficient accuracy using capillary samples.

Routine biomarkers such as glucose and HbA1c are generally suitable for capillary testing. In contrast, inflammatory cytokines (e.g., IL-6, TNF-α) and certain protein biomarkers may require more sensitive laboratory methods.

Therefore, biomarker selection should consider both biological relevance and analytical feasibility.

## Avoiding Overclaiming

Biological age models estimate physiological ageing but cannot directly measure the true biological ageing process.

Predictions generated by the enhanced model should therefore be interpreted as probabilistic estimates rather than definitive measures of health or longevity.

Clinical decisions should never rely solely on a biological age score. Instead, such models should support clinicians alongside medical history, physical examination, laboratory findings, and patient-specific risk factors.

# Conclusions

This project explored possible improvements to the original PhenoAge model by incorporating additional biomarkers, lifestyle information, and modern machine learning techniques.

The proposed framework expands the biological scope of the original model by including markers of chronic inflammation, metabolic regulation, cardiovascular health, mitochondrial function, and cellular stress.

A modular modelling strategy combining clinical biomarkers with separate lifestyle and metabolic scores may improve interpretability while reducing the risk of overfitting.

Future work should focus on validating the proposed model using large, diverse population cohorts and integrating emerging multi-omics technologies, including proteomics, metabolomics, and epigenetic biomarkers.

Overall, this enhanced framework represents a scientifically grounded conceptual extension of the original PhenoAge model and illustrates how biological age estimation could evolve with advances in precision medicine and artificial intelligence.

# Future Directions

The proposed enhanced PhenoAge framework represents one possible evolution of biological age estimation. Future developments could extend the model beyond routine clinical biomarkers by incorporating high-dimensional molecular data and continuous physiological monitoring.

Potential future extensions include:

- Proteomics to quantify circulating proteins associated with ageing.
- Metabolomics to characterise metabolic pathways and energy homeostasis.
- Epigenetic clocks based on DNA methylation.
- Transcriptomics to capture age-related changes in gene expression.
- Gut microbiome profiling.
- Continuous wearable-device monitoring.
- Artificial intelligence models capable of integrating multiple biological modalities simultaneously.

These approaches may improve the sensitivity and specificity of biological age estimation while providing greater insight into the mechanisms underlying human ageing.

- Evaluating anti-ageing therapies.
